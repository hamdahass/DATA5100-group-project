## Modeling
#### Single Input Models
We will begin with logistic regression since we are predicting a binary: malignant or benign.

###### Separate the data into training and testing sets.
X will be all of the numeric columns used to predict the diagnosis.
y will be the diagnosis column.
test_size sets the test data at 20% of the total data.
random_state = 3 makes it so that we get the same results every time. We could eliminate this later, but it could be helpful while are learning and testing our code.
stratify = y keeps the proportion of malignant/binary outcomes the same in both sets

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=['id', 'diagnosis'])
y = df['diagnosis']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=3, stratify=y
)
```

Train a model on the first predictor. Fit the model.

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train[['radius1']], y_train)
```

Check the class order so that when we look at probabilities we know whether 0 is benign and 1 is malignant or the other way around.

```python
print(model.classes_)
```

0 is benign and 1 is malignant

#### Make predictions for single input model.

```python
y_pred = model.predict(X_test[['radius1']])
y_prob = model.predict_proba(X_test[['radius1']])
```

Evaluate performance.

```python
from sklearn.metrics import accuracy_score, roc_auc_score
print("Accuracy:", accuracy_score(y_test, y_pred))
print("AUC:", roc_auc_score(y_test, y_prob[:, 1]))
```

The accuracy is 89.47%, which means that 89.47% of predictions matched the actual labels in the data set.
AUC (Area under the ROC curve) ranges from 0.5 (no better than random guessing) to 1.0 (perfect separation). The score of approximately 0.95 means the model is very good at distinguishing between malignant and benign cases.


Display a confusion matrix. Ensure that the confusion matrix is created with the correct model classes (0 is benign and 1 is malignant). Then print it with Benign and Malignant rather than 0 and 1.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Compute the confusion matrix using model.classes_ to preserve class order
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)

# Map internal class labels to human-friendly names
label_map = {'B': 'Benign', 'M': 'Malignant'}
display_labels = [label_map[label] for label in model.classes_]

# Display the confusion matrix with custom labels
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=display_labels)
disp.plot()
```

#### Multiple Input Model

The first 10 features are the mean of these categories. The second 10 features are the standard error of these categories. The final 10 features are the worst value of each of these categories.

Area and perimeter are derived from the radius. Compactness is derived from area and perimeter, thus indirectly from the radius. 

We will develop a multiple input model from the mean radius, texture, smoothness, concavity, concave_points, symmetry, and fractal_dimension, which are all independent measurements.

We will start by splitting the data into training and testing data.

```python
from sklearn.model_selection import train_test_split

X = df[['radius1', 'texture1', 'smoothness1', 'concavity1', 'concave_points1', 'symmetry1', 'fractal_dimension1']]
y = df['diagnosis']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=3, stratify=y
)
```

Then we scale the features (because radius can be over 20 and concave_points can be under 0.1) on the training set only to avoid data leakage. Then transform both training and test sets.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Then fit the multiple input logistic regression model.

```python
log_reg = LogisticRegression(max_iter = 1000)
log_reg.fit(X_train_scaled, y_train)
```

#### Make predictions on the multiple input logistic regression.

```python
y_pred = log_reg.predict(X_test_scaled)
```

#### Evaluate performance.

```python
print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Classification Report: ', classification_report(y_test, y_pred))
print('Confusion Matrix: ', confusion_matrix(y_test, y_pred))
```

The model has high accuracy and high precision, recall and f1 scores. Most cases were identified correctly as benign or malignant. 

Examine coeffieicents and multicollinearity to determine whether all the predictor variables are needed.

```python
import statsmodels.api as sm

# Add a constant term for the intercept
X_train_const = sm.add_constant(X_train_scaled.astype(float))

# # Convert B and M to 0 and 1
y_train_numeric = y_train.map({'B': 0, 'M': 1}).astype(int)

# Fit logistic regression
logit_model = sm.Logit(y_train_numeric, X_train_const).fit()
print(logit_model.summary())
```

Based on the p-values, it's possible that we don't need features x4 - x7. The model has a strong pseudo R squared value.

Check for multicollinearity by computing Variance Inflation Factors (VIFs).

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Keep scaled features as a DataFrame
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)

# Add constant for VIF calculation
X_train_const = sm.add_constant(X_train_scaled_df)

# Compute VIFs
vif_data = pd.DataFrame()
vif_data["feature"] = X_train_const.columns
vif_data["VIF"] = [variance_inflation_factor(X_train_const.values, i)
                   for i in range(X_train_const.shape[1])]
print(vif_data)
```

We were already considering keeping only radius, texture, and smoothness after checking the p-values. Check VIFs with only those 3 features included.

```python
X_subset = pd.DataFrame(X_train_scaled, columns=X_train.columns)[["radius1", "texture1", "smoothness1"]]

# Add constant
X_subset_const = sm.add_constant(X_subset)

# Compute VIFs
vif_data = pd.DataFrame()
vif_data["feature"] = X_subset_const.columns
vif_data["VIF"] = [variance_inflation_factor(X_subset_const.values, i)
                   for i in range(X_subset_const.shape[1])]
print(vif_data)
```

```python
from sklearn.metrics import accuracy_score, roc_auc_score

y_pred = rf.predict(X_test)
y_prob = rf.predict_proba(X_test)[:, 1]  # probability of malignant

print("Accuracy:", accuracy_score(y_test, y_pred))
print("AUC:", roc_auc_score((y_test=='M').astype(int), y_prob))
```

## XGBoost Model
Use all predictor variables and let XGBoost determine which to keep.

Check whether the target variable is skewed.

```python
df['diagnosis'].value_counts()
```

This is about a 63/37 split, so I'm going to proceed with the model. 

Change the M/B in the diagnosis column to 1/0 in the target series. 
Malignant: 1
Benign: 0

```python
y = df['diagnosis'].map({'M': 1, 'B': 0})
```

Create the feature matrix.

```python

X = df.drop(columns=['id', 'diagnosis'])
```

Split the data into training and testing data.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=3, stratify=y
)
```

```python
from sklearn.model_selection import GridSearchCV
import xgboost as xgb

# Base model
xgb_clf = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Parameter grid
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 4, 5],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Grid search with cross-validation
grid_search = GridSearchCV(
    estimator=xgb_clf,
    param_grid=param_grid,
    scoring='accuracy',
    cv=5,
    verbose=1
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best CV accuracy:", grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_
```

Check the accuracy with all the features.

```python
# Predict on the test set
y_pred_full = best_model.predict(X_test)

# Accuracy with all features
baseline_accuracy = accuracy_score(y_test, y_pred_full)
print("Baseline accuracy with all features:", baseline_accuracy)
```

Plot feature importances.

```python

from xgboost import plot_importance

plt.figure(figsize=(10, 6))
plot_importance(best_model, max_num_features=10, importance_type='gain')
plt.title("Top 10 Feature Importances (by gain)")
plt.show()
```
